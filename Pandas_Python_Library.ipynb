## (ii) Basics of pandas

### (a) `Series`

A Series is a one-dimensional labeled array. It can hold any data type (integers, strings, floats, etc.).

#### Syntax

```python
import pandas as pd
s=pd.Series([10, 20, 30])
```

### (b) `DataFrame`

A DataFrame is a two-dimensional labeled data structure (like a table with rows and columns).

#### Syntax

```python
df=pd.DataFrame({
    "Name":["Ali","Saqib"],
    "Age":[20,22]
})
```

### (c) `dtypes`

dtypes function is used to check data type of the whole data frame.

#### Syntax

```python
print(df.dtypes)
```

### (d) `dtype`

dtype function is used to check data type of a specific column

#### Syntax

```python

```

### (e) `index`

The index is like the label of rows in Series or DataFrame. It helps in locating data.

#### Syntax

```python
print(df.index)  # gives the index range (e.g., 0, 1, 2, ...)
```

### (f) `columns`

Columns are the labels for the vertical data fields in a DataFrame.

#### Syntax

```python
df.columns  # shows column names: ['Name', 'Age']
```

### (g) `loc[]`

The loc[] function is used to access data by label (index or column name).

#### Syntax

```python

print(data_frame.loc[0])  # returns the first row
```

### (h) `iloc[]`

The iloc[] function is used to access data by position (row or column number).

#### Syntax

```python

```

### (i) `head()` and `tail()`

### (j) `size`

### (k) `shape`

## (iii) unicodedecodeerror

The UnicodeDecodeError in pandas usually happens when you're trying to read a file (like a CSV) that contains special or non-English characters, but Python doesn't know how to interpret those characters properly. 

**Example**

```python
import pandas as pd
df = pd.read_csv("data.csv")
```

* UnicodeDecodeError: 'utf-8' codec can't decode byte 0x96 in position 10: invalid start byte

* This means pandas tried to read the file as UTF-8, but the file uses a different encoding, like ISO-8859-1, latin1, or cp1252.

### (a) UTF-8

UTF-8 stands for "Unicode Transformation Format – 8-bit", and it's the most common character encoding used in computers and the internet today.

* It’s a way to store and represent text characters (like English, Urdu, Chinese,) as bytes (0s and 1s).

* We can fix this error using encoding parameter.

```python
df = pd.read_csv("data.csv", encoding='latin1')  # or 'ISO-8859-1' or 'cp1252'
```

If it is still not working than use.

```python
df = pd.read_csv("data.csv", encoding='unicode_escape')
```

## (vi) Selecting and Filtering

In pandas, selecting and filtering data allows you to extract rows and columns based on specific conditions or structure.

### (a) Selecting Columns

Selecting columns means accessing one or more specific columns (features) from a DataFrame. This is useful when you want to analyze, manipulate, or visualize only part of your data.

#### Syntax

Selecting single column:-

```python
df["Column1"]
```

Selecting multiple columns:-

```python
df[["Column1", "Column2","columnN"]]
```

### (b) Selecting rows 

Selecting rows means accessing one or more specific records (rows) from a DataFrame, usually based on their index. This is essential when you want to view, filter, or operate on specific entries in your dataset.

#### Syntax

* Using loc[] (Label-based selection)

* Used when you want to select rows by index labels.

```python
df.loc[2]           # Single row by index label
```

```python
df.loc[1:3]         # Multiple rows using index label range (inclusive)
```

```python
df.loc[[0, 2]]      # Specific multiple rows by label
```

* Using iloc[] (Integer position-based selection)

* Used when you want to select rows by integer index.

```python
df.iloc[2]          # Single row by position
```

```python
df.iloc[1:3]        # Multiple rows (like list slicing, exclusive of end)
```

```python
df.iloc[[0, 2]]     # Specific multiple rows by position
```

### (c) Filtering Rows by Condition

Filtering rows means selecting only those rows from a DataFrame that meet a specific condition or set of conditions. This is often used to extract meaningful subsets of data based on values in one or more columns.

#### Syntax

```python
df[ df['ColumnName'] condition ] # for single condition
```

```python
df[ (df['ColumnName'] condition) operator (df['ColumnName'] condition) ... ] # for multiple condition.
```

### (d) Using `.query()` for Filtering

.query() lets you filter rows using a string expression that looks like a condition you'd write in SQL or plain math logic. It’s great for avoiding a lot of brackets and df['col'] syntax.

* Column names with spaces must be enclosed in backticks (`) in df.query().

#### Syntax(For Selecting all columns)

```python
df.query("condition")
```

#### Syntax(For selecting specific columns)

```python
df.query("condition")[['col1', 'col2']]
```

### (e) Filtering with `.isin()` or `.str`

#### `.isin()`

.isin() is used to filter rows where a column's value matches any value in a list.

##### Syntax

```python
df[df['Column'].isin([value1, value2, ...])]
```

## (vii) Data Modification in pandas

Data modification in pandas involves changing or updating the data in your DataFrame. It includes adding, updating, deleting rows or columns, or changing individual values.

## (viii) Handling Missing Data

Missing values refer to the absence of data or information in a dataset where a value is expected. These are entries in the data that are blank, null, or not recorded, and they can occur for a variety of reasons.

### (a) Common missing values

NaN (Not a Number), None, null, Empty strings (""), A special value like -9999.

#### How to check our data contains missing values:-

##### `isnull()`

The isnull() method returns a data frame of boolean values. Returns True if data conatins missing value and returns False if value don't contains missing values.

**Syntax**

```python
print(data_frame.isnull())
```

##### `isnull().sum()`

The isnull().sum() counts how many missing values we have in each column.

**Syntax**

```python
print(data_frame.isnull().sum())
```

#### How to remove missing values:-

We can remove missing values using dropna()

**Syntax**

```python
data_frame.dropna(inplace=True)
```

#### How fill the misiing values with values.

We can fill missing values using fillna()

**Syntax**

```python
data_frame.fillna(value,inplace=True) # Filling all missing values with a single value
```

```python
data_frame["Column_Name"].fillna(value,inplace=True) # Filling missing values in a specific column
```
#### Interpolation:-

Interpolation is a technique used to fill missing values in a dataset by estimating them based on the existing values.   

* Instead of just filling missing data with a fixed value (like mean or median), interpolation uses patterns in the data (like increasing or decreasing trends) to estimate the missing values more accurately.

**Syntax**

```python
df.interpolate(method='linear', axis=0, inplace=False,limit_direction='forward')
```

| Method | Description |
|---|---|
| `'forward'` | Fill from earlier values downward (default for most cases) |
| `'backward'` | Fill using later values upward |
| `'both'` | Use both forward and backward interpolation (from either side) |

##### Common Interpolation Methods in Pandas

| Method | Description |
|---|---|
| `'linear'` | Default method. Assumes values change linearly between points |
| `'time'` | Used when the index is datetime; useful for time series |
| `'index'` | Uses the index values for interpolation |

## (xi) Crosstabs and pivote table.

### (a) Crosstabs

crosstab in Pandas is a function used to compute a cross-tabulation of two or more factors. It is mainly used to analyze the relationship between two categorical variables — like making a frequency table.

**Syntax**

```python
pd.crosstab(index, columns, values=None, aggfunc=None, margins=False)
```

| Parameter | Description |
|---|---|
| `index` | Rows (usually one or more categorical variables) |
| `columns` | Columns (another categorical variable) |
| `values` | Optional numeric values to aggregate |
| `aggfunc` | Aggregation function such as `sum`, `mean`, etc. |
| `margins` | Adds totals (row/column sums) if `True` |

* Designed mainly for categorical variables

### (b) Pivote Tables

A pivot table in Pandas is a powerful tool to summarize, group, and analyze your data — especially useful when working with large datasets.

**Syntax**

```python
pd.pivot_table(data, index=None, columns=None, values=None, aggfunc='mean')
```

| Parameter | Description |
|---|---|
| `data` | The DataFrame |
| `index` | Column(s) to group by (rows) |
| `columns` | Column(s) to group by (columns) |
| `values` | Column to aggregate |
| `aggfunc` | Aggregation function (default is `'mean'`; others include `sum`, `count`, `min`, `max`) |

* Works with both categorical and numerical data.


````python
import pandas as pd
# Sample dataset
data={
    'Region': ['East', 'East', 'West', 'West', 'East', 'West'],
    'Category': ['Tech', 'Furniture', 'Tech', 'Tech', 'Furniture', 'Furniture'],
    'Sales': [200, 150, 300, 400, 100, 250]
}
df=pd.DataFrame(data)
print("Original DataFrame")
print(df)

# 1. Basic Frequency Count (How many sales occurred per Region and Category?)
freq_table=pd.crosstab(index=df['Region'], columns=df['Category'], margins=True)

print("Crosstab: Frequency Count")
print(freq_table)

print("\n"+"="*40+"\n")

# 2. Aggregation with values (Total Sales amount per Region and Category)
sum_crosstab=pd.crosstab(
    index=df['Region'], 
    columns=df['Category'], 
    values=df['Sales'], 
    aggfunc='sum', 
    margins=True
)

print("Crosstab: Total Sales Sum")
print(sum_crosstab)

# 1. Default Aggregation (Average Sales per Region and Category)
pivot_mean = pd.pivot_table(
    data=df, 
    index='Region', 
    columns='Category', 
    values='Sales'  # Default aggfunc='mean'
)

print("Pivot Table: Average Sales")
print(pivot_mean)

print("\n"+"="*40+"\n")

# 2. Custom Aggregation (Total Sales Sum with Margins)
pivot_sum=pd.pivot_table(
    data=df, 
    index='Region', 
    columns='Category', 
    values='Sales', 
    aggfunc='sum', 
    margins=True
)

print("Pivot Table: Total Sales Sum")
print(pivot_sum)
````
