# Matplotlib

## (i) What is Data Visualization

Data Visualization is the graphical representation of information and data using visual elements like:

* Charts (bar chart, pie chart, line chart), Graphs, Maps, Plots (scatter plots, box plots, etc.)

* It turns complex data into a visual format that makes it easier to Understand trends, Identify patterns, Detect outliers, Communicate insights clearly.

## (ii) Why Do We Use Data Visualization

We use data visualization because:

### (a) Simplifies Complex Data

It helps make large or complex datasets easier to understand.

### (b) Quick Insights

Visuals are faster to process than raw numbers or tables.

### (c) Better Decision-Making

Clear visuals help businesses and data analysts make smarter decisions.

### (d) Pattern & Trend Recognition

You can easily spot trends, outliers, and relationships in data.

### (e) Engaging and Interactive

Visuals are more attractive and often interactive, which helps keep attention.

### (f) Helps in Storytelling

Makes it easier to tell a data-driven story to stakeholders.

#### Example

Suppose you have sales data for 12 months.

**Without visualization** You’ll look at 12 numbers in a table and try to figure out which month had the highest/lowest sales.

**With visualization (like a line chart)** You can immediately see the sales trend across the year.

## (v) Understanding pyplot

The pyplot is a module in the Matplotlib library used for creating visualizations like Line plots, Bar charts, Histograms, Pie charts, Scatter plots.

* It provides a simple way to build graphs step-by-step, like drawing with commands.

* Think of pyplot like a drawing canvas.

* We tell it:

    * What to draw (plot, bar, scatter), How to style it (color, marker, line), What labels/titles to show (xlabel, ylabel, title), Then display it using plt.show()

### (a) Why we use pyplot

Easy to learn, Step-by-step plotting, Works well with NumPy and Pandas, Can create complex visualizations with a few lines of code.

## (vi) Key Terms you should know

### (a) x-axis and y-axis

* X-Axis (Horizontal Axis) goes left to right. It represents the independent variable. You usually put things like time, categories, or labels on the X-axis.

* Y-Axis (Vertical Axis) goes bottom to top. It represents the dependent variable. You usually plot quantities, values, or measurements on the Y-axis.

### (b) Data point

A data point in Matplotlib represents a single piece of data plotted on a graph. It usually consists of two values( X-coordinate (horizontal position), Y-coordinate (vertical position))

#### Example

```python
X = [1, 2, 3, 4]
Y = [10, 20, 25, 30]
```
* Each pair like (1,10), (2,20) is a data point.

### (c) Figure

In Matplotlib, a Figure is the entire window or page where your plots, charts, axes, titles, and labels appear.

* Think of a Figure as a blank canvas that can hold:

* One or more plots (also called Axes), Titles, Legends, Grid lines, Colorbars and other elements.


### (d) Axes

In Matplotlib, Axes is the actual plot or graph where your data is visualized. It includes x-axis, y-axis, The data itself (lines, bars, scatter points, etc.) and also titles, gridlines, tick marks, legends, etc.

* In simple words A Figure is like a paper sheet and an Axes (plot) is like a drawing on that paper.

### (e) Plot

A plot is the visual representation of data on an Axes (graph) in Matplotlib.

* It shows how values relate to each other, often by using, Lines, Markers, Bars Points, etc.

### (f) Marker

A marker is a symbol used to represent individual data points on a plot.

* By default, a line plot just connects points with lines. If you add markers, each data point becomes visible as a dot, circle, square, etc.

### (g) Line and Line Style

* In Matplotlib, a line connects data points on a graph — usually in a line plot.

* Line style means the appearance of the line — like whether it’s: Solid("-" By default), Dotted(":"), Dashed("--"), Dash-dotted("-.")

### (h) Color

In Matplotlib, color is used to change the appearance of lines, markers, bars, Text, Axes, backgrounds.

* Short Color Codes:

'b'----Blue, 'g'----Green, 'r'----Red, 'c'----Cyan, 'm'----Magenta, 'y'----Yellow, 'k'----Black, 'w'----White

### (i) Legend

A legend is a box that explains the meaning of different lines, colors, or markers on your plot.

* It tells the viewer: “This red line is Sales, this blue line is Profit,” etc.

#### Syntax


## (viii) Bar chart

A bar chart is a graph that uses rectangular bars to represent data values, where the length or height of the bar shows the value of the category.

* Each bar represents a category.

* The x-axis usually holds categories, and the y-axis holds values (or vice versa for horizontal bars).

* Bars can be vertical or horizontal.

### (a) Why We Use Bar Charts

* Compare Categories Easily.

* Works with Categorical Data.

* Handle Large Value Differences.

* Easy to Interpret.

### (b) Plotting Bar charts

#### `plt.bar()`

plt.bar() in Matplotlib is used to make bar charts (vertical bars). It’s perfect when you want to compare categories or show discrete data.

**Syntax**

```python
plt.bar(x, height, width=0.8, color=None, edgecolor=None, linewidth=None,align="Center")
```

| Parameter | Description |
|---|---|
| `align` | `'center'` (default) or `'edge'`; specifies how bars are aligned to the x positions |

````python
import matplotlib.pyplot as plt
months=["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]
saled_items=[59,43,35,65,67,51,30,49,86,66,37,77]
plt.bar(months,saled_items,color="Orange",label="Saled Item")
plt.xlabel("Months")
plt.ylabel("Saled Items")
plt.title("Sales Per Month")
plt.ylim(10,100)
plt.legend(loc="upper right")
plt.minorticks_on()
plt.show()
plt.barh(months,saled_items,color="Orange",label="Saled Item")
plt.xlabel("Saled Items")
plt.ylabel("Months")
plt.title("Sales Per Month")
plt.legend(loc="lower right")
plt.minorticks_on()
plt.show()
````

## (ix) Pie chart

A pie chart is a circular chart that’s divided into slices, where each slice represents a proportion of the whole.

-> A pie chart visually shows parts of a whole as percentages or fractions, with The whole circle = 100% (or the total sum of your data), Each slice = a category’s contribution to the total

### (a) Why We Use Pie Charts

* Show Proportions Clearly.

* Highlight the Largest or Smallest Share.

* Works best when there are few categories (ideally 5–6 max).

* Easy to Understand for Non-Technical Audiences.

### (b) Plotting Pie chart

#### `plt.pie()`

plt.pie() in Matplotlib is used to create a pie chart, which shows how a whole is divided into different parts.

**Syntax**

```python
plt.pie(x, labels=None, colors=None, autopct=None, startangle=0, shadow=False, explode=None)
```

**Main Parameters**

| Parameter | Description |
|---|---|
| `x` | List/array of values representing the sizes of each wedge |
| `labels` | Names for each wedge |
| `colors` | List of colors for the wedges |
| `autopct` | Format string or function to display percentages (e.g., `'%1.1f%%'`) |
| `startangle` | Rotates the chart by the given degrees (e.g., `90` to start from the top) |
| `shadow` | If `True`, adds a shadow effect |
| `explode` | List used to separate wedges from the center (e.g., `[0, 0.1, 0, 0]`) |

````python
import matplotlib.pyplot as plt
regions=["North","South","East","West"]
revenue=[40000,50000,60000,25000]
plt.figure(figsize=(5,9))
plt.pie(revenue,labels=regions,colors=["Red","Yellow","Green","Blue"],autopct="%1.1f%%")
plt.title("Revenue contribution by Region")
plt.legend()
plt.tight_layout()
plt.show()
water_and_land=[71,29]
plt.pie(water_and_land,labels=["Water","Land"],colors=["dodgerblue","limegreen"])
plt.title("Water and Land on Earth")
plt.legend()
plt.show()
````

## (x) Histograms

A histogram is a type of graph that shows the distribution of a dataset — in other words, it tells you how often values fall within certain ranges (called bins).

* Bars that touch each other (unlike bar charts with gaps).

* The x-axis shows value ranges (e.g., 0–10, 10–20, 20–30).

* The y-axis shows frequency (how many values are in each range).

### (a) Why We Use Histograms

* Understand Data Distribution.

* Spot Patterns and Anomalies.

* Summarize Large Data Quickly.

* Compare Data Spread.

### (b) Plotting Histogram chart

#### `plt.hist()`

plt.hist() in Matplotlib is the function we use to create a histogram — it plots the distribution of a dataset by dividing values into bins and showing how many data points fall into each bin.

**Syntax**

```python
plt.hist(data,bins=None,range=None,color=None,edgecolor=None)
```

| Parameter | Description |
|---|---|
| `x` | The data you want to plot |
| `bins` | Number of intervals or specific bin edges |
| `range` | The minimum and maximum values for the histogram |
| `color` | Fill color of the bars |
| `edgecolor` | Color of the bar borders |

````python
import matplotlib.pyplot as plt
students_marks=[65,87,34,55,90,98,76,23,17,80,99,100,67,52,77,100,79,66,60,40]
plt.hist(x=students_marks,bins=8,color="Red",edgecolor="Black")
plt.xlabel("Marks Range")
plt.ylabel("Number of Students")
plt.xlim(0,100)
plt.ylim(0,6)
plt.title("Marks Comparison")
plt.minorticks_on()
plt.show()
````

## (xi) Scatter Plot

A scatter plot is a type of graph that displays individual data points on a two-dimensional plane, where, The x-axis represents one variable, The y-axis represents another variable, Each point on the plot shows the value of both variables for a single observation.

### (a) Why we use it

* To visualize the relationship or correlation between two variables.

* To identify patterns, trends, or clusters in data.

* To detect outliers (points that don’t follow the general pattern).

### (b) Plotting Scatter Plot

#### `plt.scatter()`

plt.scatter() is a Matplotlib function used to create scatter plots in Python.

**Syntax**

```python
plt.scatter(x,y,size=None,color=None,marker=None,label="label Name")
```

| Parameter | Description |
|---|---|
| `x` | Values for the x-axis (list, array, or Series) |
| `y` | Values for the y-axis |
| `s` | Size of points (default is `20`) |
| `c` | Color of points (name, hex code, or sequence of colors) |

````python
import matplotlib.pyplot as plt
temperature=[30,35,29,37,24,40,39]
ice_cream_sales=[50,70,40,80,25,90,85]
plt.scatter(temperature,ice_cream_sales,color="Red",marker="o",label="Data point(Temperature,Ice cream sales)")
plt.xlabel("Temperture")
plt.ylabel("Ice cream Sales")
plt.title("Relationship between Temperatue and Ice cream")
plt.grid()
plt.legend()
plt.minorticks_on()
plt.show()
````

## (xii) Stem plot

A stem plot (also called stem-and-leaf plot) is a way of organizing numerical data to see its shape (distribution) while keeping the actual data values.

### (a) Why do we use Stem Plots

**Quick visualization**

Helps to see how the data is spread out (like small histogram).

**Keeps original data**

Unlike bar charts or histograms, it does not lose the actual values.

**Easy to create** 

Good for small datasets.

**Shows distribution** 

Lets us see clusters, gaps, and outliers clearly.

### (b) Plotting Stem Plot

#### `plt.stem():-`

plt.stem() is a Matplotlib function used to create stem plot in Python.

**Syntax**

```python
plt.stem(x, y, linefmt="color_first_letter no_space linestyle", markerfmt="color_first_letter no_space marker", basefmt= "color_first_letter no_space linestyle", bottom(place of base line)=value, orientation="vertical/horizontal",label="label")
```

* If we want to customize the appearance of different elements in a stem plot, we can use the objects returned by `plt.stem()`.

```python
markerline, stemlines, baseline = plt.stem(
    x1,
    y1,
    bottom=value,
    orientation="vertical",
    label="label"
)

markerline.set_markersize(size)
markerline.set_markerfacecolor("marker color")
markerline.set_markeredgecolor("marker edge color")

stemlines.set_color("color")
stemlines.set_linewidth(size)

baseline.set_color("Baseline color")
baseline.set_linewidth(size)
baseline.set_linestyle("linestyle")
```

````python
import matplotlib.pyplot as plt
x1=[1,3,5,7,9]
y1=[2,4,6,8,10]
plt.stem(x1,y1,markerfmt="bo",linefmt="b--",bottom=6,basefmt="g-",label="Data point",orientation="horizontal")
plt.show() 
markerline,stemlines,baseline=plt.stem(x1,y1,bottom=0,orientation="vertical",label="Data point")
markerline.set_markersize(8)
markerline.set_markerfacecolor("Red")
markerline.set_markeredgecolor("Black")
stemlines.set_color("Gray")
stemlines.set_linewidth(4)
stemlines.set_linestyle("--")
baseline.set_color("Black")  
baseline.set_linewidth(2)      
baseline.set_linestyle("-")
plt.show()
````

## (xiii) Step plot

A step plot is a type of line plot where data points are connected with horizontal and vertical lines instead of diagonal straight lines. It looks like a series of "steps" rather than slanted lines.

* In Matplotlib, we usually make it with `plt.step()` instead of `plt.plot()`.

### (a) Why we use Step Plot

* We use a step plot when:

    * Data changes at specific intervals – For example, electricity meter readings, stock price updates, or digital signals.

    * To show sudden jumps or constant values between points – Unlike line plots that imply smooth change, step plots show that the value stays constant until the next change.

    * Better representation of discrete data – Useful for categorical x-values or piecewise constant functions.

### (b) Plotting Step Plot

#### `plt.step():-`

plt.step() is a Matplotlib function used to create step plot in Python.

**Syntax**

```python
plt.step(data=data, x="x-axis", y="y-axis", color="Wcolor", marker="marker", linestyle="linstyle",linewidth=value,mfc="color", ms=value, mew=value, label="label",alpha=value, where="pre/mid/post")
```

````python
import matplotlib.pyplot as plt
x2=[1,3,5,7,9]
y2=[2,4,6,8,10]
plt.step(x=x2,y=y2,color="black",marker="o",linestyle="--",linewidth=4,mfc="Red",ms=10,mew=2,label="Data point",alpha=0.9,where="mid")
plt.grid()
plt.legend()
plt.show()
````

## (xiv) Stack or Area Plot

An area plot (or stack plot) is a line plot where the area beneath the line is filled with color. When multiple data series are plotted together, they are stacked on top of one another to show how individual categories contribute to a total sum over time or across intervals.

### (a) Why we use a Stack plot

* We use a stack plot when:

    * Showing part-to-whole relationships over time – For example, displaying total company revenue broken down by different products.

    * Tracking cumulative trends – Useful for seeing both individual category changes and the overall total growth simultaneously.

    * Visualizing proportion changes – Helps highlight how the composition of a dataset shifts over discrete steps or time periods.

**Syntax**

```python
plt.stackplot(x, y1, y2, y3, labels=['Label 1', 'Label 2'], colors=['color1', 'color2'], alpha=value, baseline='zero')
```

**Parameters**

| Parameter | Description |
|---|---|
| `x` | X-axis values (usually continuous, such as time, days, or months) |
| `y1, y2, ...` | Arrays/Series of numerical values representing each layer to stack |
| `labels` | List of string labels for the legend corresponding to each series |
| `colors` | List of color codes for filling the areas beneath each layer |
| `alpha` | Sets the transparency level of the filled areas (`0.0` to `1.0`) |
| `baseline` | Determines how the layers are stacked: `'zero'`, `'sym'`, or `'wiggle'` |

````python
import matplotlib.pyplot as plt
months=["Jan","Feb","Mar","Apr","May"]
online_sales=[10,15,20,25,30]
in_store_sales=[20,18,15,12,10]
plt.stackplot(months,online_sales,in_store_sales,labels=["Online Sales","In-Store Sales"],colors=["#1f77b4","#ff7f0e"],alpha=0.8)
plt.title("Monthly Sales Breakdown")
plt.xlabel("Month")
plt.ylabel("Total Units Sold")
plt.legend(loc="upper left")
plt.show()
````

## (xv) Subplots and Layout Adjustments

A subplot in Matplotlib is a way to display multiple plots within a single figure. Instead of showing only one chart, you can divide the figure into a grid and place a different plot in each section.

### (a) Why we use subplots

* To compare different datasets or visualizations side-by-side.

* To save space by combining plots in one figure.

* To create dashboards or multi-chart reports.

### (b) We can create subplots in two ways by using `plt.subplot()` and using `plt.subplots()`

#### `plt.subplot:-`

plt.subplot() is a Matplotlib function used to create a single subplot inside a grid layout within one figure.

* It divides the figure into rows and columns, then selects one position to draw the plot.

**Syntax**

```python
plt.subplot(nrows, ncols, index) # In this context index starts from 1.
```

* In a single plt.subplot() figure, the number of rows and columns stays the same for all the subplots in that figure.

#### Figure (Object oriented way(Most useful and recomended))

Matplotlib’s object-oriented interface, we explicitly create a Figure and Axes objects using plt.subplots(), then call methods on those Axes instead of using the state-based plt functions.

**plt.subplots**

plt.subplots() in Matplotlib is a figure and axes creation function that, Creates a Figure(the whole canvas), Creates one or more Axes objects(individual plotting areas), Returns both so you can use them in the object-oriented interface.

##### Syntax

```python
fig, ax = plt.subplots(nrows=1, ncols=1, figsize=None)
```

| Argument | Description |
|---|---|
| `rows` | Number of rows of subplots (default: `1`) |
| `ncols` | Number of columns of subplots (default: `1`) |
| `figsize` | Tuple `(width, height)` in inches for the figure size |

| Return Value | Description |
|---|---|
| `fig` | The `Figure` object (overall container) |

## (xvi) Reference Lines and annotations

### (a) Reference Lines

In Matplotlib, Reference Lines are tools to highlight specific points, thresholds, or regions in your plots so your data story becomes clearer.

#### `plt.axhline()` Horizontal Reference Line

Draws a straight horizontal line across the Axes.

**Syntax**

```python
plt.axhline(y, color='r', linestyle='--', linewidth=2)
```

| Parameter | Description |
|---|---|
| `y` | Y-position of the horizontal line |

| `linestyle` | Style of the line (e.g., `'-'`, `'--'`, `':'`) |

#### `plt.axvline()` (Vertical Reference Line)

Draws a straight vertical line.

**Syntax**

```python
plt.axvline(x, color='g', linestyle=':')
```

| Parameter | Description |
|---|---|
| `x` | X-position of the vertical line |

* Style parameters are the same as ax.axhline().

#### `plt.axhspan()`

plt.axhspan() in Matplotlib draws a horizontal shaded band across the plot between two Y-axis values, useful for highlighting ranges like thresholds or confidence intervals.

**Syntax**

```python
plt.axhspan(ymin, ymax, xmin=0, xmax=1, color=None, alpha=None, **kwargs)
```

| Parameter | Description |
|---|---|

#### `plt.axvspan()`


| Parameter | Description |
|---|---|

| `**kwargs` | Additional style settings |

### (b) Annotation

A note by a way of explanation or comment added to a text or diagram

#### `plt.annotate()`

plt.annotate() in Matplotlib is used to add text annotations to a specific point on a plot, often with an optional arrow pointing to it. It’s great for highlighting important data points or adding context.

**Syntax**

```python
plt.annotate(text, xy, xytext=None, arrowprops=None, **kwargs)
```

| Parameter | Description |
|---|---|
| `text` | The annotation text (string) |
| `xy` | Coordinates `(x, y)` of the point you want to annotate |
| `xytext` | Coordinates `(x, y)` where the text should be placed (optional; defaults to `xy`) |
| `arrowprops` | Dictionary used to customize the arrow style |
| `**kwargs` | Other text formatting options such as `fontsize`, `color`, etc. |

````python
import matplotlib.pyplot as plt
import numpy  as np
from statistics import mode
x_axis_data=[15,24,12,16,25,27,36,21,34,23]
y_axis_data=[5,50,15,45,25,35,26,30,20,10]
x_axis_mean=np.mean(x_axis_data)
print(f"x_axis_mean: {x_axis_mean}")
x_axis_median=np.median(x_axis_data)
print(f"x_axis_median: {x_axis_median}")
x_axis_mode=mode(x_axis_data)
print(f"x_axis_mode: {x_axis_mode}")
y_axis_mean=np.mean(y_axis_data)
print(f"xy_axis_mean: {y_axis_mean}")
y_axis_median=np.median(y_axis_data)
print(f"y_axis_median: {y_axis_median}")
y_axis_mode=mode(y_axis_data)
print(f"y_axis_mode: {y_axis_mode}")
plt.plot(x_axis_data,y_axis_data,color="Black",marker="o",label="Data point (x,y)")
plt.axvline(x_axis_mean,color="Red",label="x-axis mean")
plt.axvline(x_axis_median,color="Blue",label="x-axis median")
plt.axvline(x_axis_mode,color="Green",label="x-axis mode")
plt.axhline(y_axis_mean,color="Orange",label="y-axis mean")
plt.axhline(y_axis_median,color="Yellow",label="y-axis median")
plt.axhline(y_axis_mode,color="Skyblue",label="y-axis mode")
plt.xlabel("Values of x")
plt.ylabel("Values of y")
plt.title("Comparison between Values of x and Values of y")
plt.grid()
plt.minorticks_on()
plt.legend(loc="upper right",fontsize=9)
plt.annotate("(x=34,y=20)",(34,20),(31,11),arrowprops={"arrowstyle":"->","color":"Red"},color="Black")
# plt.savefig("Graph3.png",dpi=300,bbox_inches="tight")
plt.show()
````
